### Missing Value Handling

```python
df.dtypes
#No object datatype beacuse in Python 3.14.6 it is changed to str datatype.
```

```python
#Replacing missing categorical values with mode and numeric values with median.
for col in df.columns:
    if str(df[col].dtype) == 'str':
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())

print(df.isnull().sum().sum())
```

### Scaling

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

num_cols = df.select_dtypes(include=np.number).columns

standard_df = df.copy()
minmax_df = df.copy()

standard_df[num_cols] = StandardScaler().fit_transform(standard_df[num_cols])
minmax_df[num_cols] = MinMaxScaler().fit_transform(minmax_df[num_cols])
```

```python
standard_df[num_cols].head()
```

```python
minmax_df[num_cols].head()
```

Standard Scaling: Best for algorithms such as Logistic Regression, SVM, and PCA because it centers data around zero with unit variance.

Min-Max Scaling: Preferred for Neural Networks and distance-based algorithms when values should remain within a fixed range (0 to 1).

### Encoding

```python
#Seeing how many unique categories in each categorical column
df.select_dtypes(include='str').nunique()
```

```python
from sklearn.preprocessing import LabelEncoder

sample_df = df.copy()

sc = [] #Small categorical columns for One-Hot Encoding
lc = [] #Large categorical columns for Label Encoding

for col in sample_df.columns:
    if str(sample_df[col].dtype) == 'str':
        if sample_df[col].nunique() < 5:
            sc.append(col)
        else:
            lc.append(col)

print("Columns for One-Hot Encoding:", sc)
print("Columns for Label Encoding:", lc)
```

```python
# One-Hot Encoding
encoded_df = pd.get_dummies(sample_df, columns=sc)

# Label Encoding
le = LabelEncoder()
for col in lc:
    encoded_df[col] = le.fit_transform(encoded_df[col])

encoded_df.head()
```

**One-Hot Encoding**
- No false ordering
- Increases dimensions

**Label Encoding**
- Memory efficient
- Can introduce artificial ordering

### Outlier Removal using Isolation Forest

```python
from sklearn.ensemble import IsolationForest

iso = IsolationForest(contamination=0.05, random_state=42)
pred = iso.fit_predict(df[num_cols])
#1 is not outlier adn -1 is a outlier
clean_df = df[pred == 1]
print('Original:', len(df))
print('After removing outliers:', len(clean_df))
```

Outliers may distort model training, increase variance, and reduce prediction accuracy.

### PPS Score and Correlation

**Comparison**

- PPS detects linear and non-linear predictive relationships and also works with categorical variables.
- PPS is generally more informative for feature selection in mixed datasets.
