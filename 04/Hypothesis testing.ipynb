```python
from scipy.stats import norm
```

```python
#Given values
sample_mean =3050
mean_units =600

#Theoretical weekly operating cost
#Using the forlmula W = 1000 + 5X
theoretical_mean = 1000+(5*mean_units)

sigma =5*25 #Sigma value
n =25 #Sample size

alpha =0.05 #Significance level
```

```python
print("Null Hypothesis (H0):")
print("The average weekly operating cost is equal to the theoretical cost.")

print("\nAlternative Hypothesis (H1):")
print("The average weekly operating cost is higher than the theoretical cost.")
```

```python
# Calculate Z-test statistic
#Formula= (Sample mean-population mean)/(Sigma/sqrt(n))
z = (sample_mean-theoretical_mean)/(sigma/(n**0.5))
print("Test Statistic (Z) =", round(z, 2))
```

```python
#Finding the critical value for the given significance level
critical_value = norm.ppf(1 - alpha)
print("Critical Value =", round(critical_value, 2))
```

```python
#Comparing the test statistic with the critical value to make a decision
if z > critical_value:
    print("Decision: Reject the Null Hypothesis")
else:
    print("Decision: Fail to Reject the Null Hypothesis")
```

```python
#Printing the final conclusion based on the decision
if z > critical_value:
    print("Conclusion:")
    print("There is enough evidence to support the restaurant owners' claim that the weekly operating costs are higher.")
else:
    print("Conclusion:")
    print("There is not enough evidence to support the restaurant owners' claim.")
```
