## Exploratory Data Analysis on Cardiographic Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#Loading the dataset
df = pd.read_csv("cardiotocographic.csv")

#Displaying first 5 rows
df.head()
```

```python
#Checking the size of the dataset
print("Rows and Columns:", df.shape)

#Checking column names
print(df.columns)

#Basic information
df.info()
```

```python
#Checking data types of each column
print(df.dtypes)
```

```python
#Checking for missing values
print(df.isnull().sum())

#If missing values exist, replace them with the column mean
df = df.fillna(df.mean())
```

```python
#Statistical summary
df.describe()
```

```python
#Mean 
print("Mean")
print(df.mean())

#Median
print("\nMedian")
print(df.median())

#IQR
print("\nInterquartile Range (IQR)")
iqr = df.quantile(0.75)-df.quantile(0.25)
print(iqr)
```

```python
plt.figure(figsize=(12,12))

for i, col in enumerate(df.columns,1): #1 because we want to start the index from 1 for subplotting
    plt.subplot(5,3,i)  #Start from position 1
    sns.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

* LB has few outliers but most of data is concentrated between 125 and 140.
* AC has several outliers with half of data lying between 0.0 to 0.005
* FM: FM has many outliers above the upper whisker, while most of the data is concentrated near 0.
* UC: UC has a few outliers, with most of the data lying between approximately 0.002 and 0.006.
* DL: DL has several outliers above the upper whisker, while most values are clustered close to 0.
* DS: DS has very few non-zero values that appear as outliers, with almost all observations equal to 0.
* DP: DP has very few non-zero values that appear as outliers, with almost all observations equal to 0.
* ASTV: ASTV has one lower outlier, while most of the data is concentrated between approximately 30 and 60.
* MSTV: MSTV has one lower outlier, with most values lying between approximately 0.8 and 1.8.
* ALTV: ALTV has many upper outliers and one lower outlier, while most of the data is concentrated below 10.
* MLTV: MLTV has several upper outliers along with one lower outlier, with most values lying between approximately 5 and 10.
* Width: Width has one upper outlier and one lower outlier, while most of the data is concentrated between approximately 40 and 90.
* Tendency: Tendency has one upper and one lower outlier, with most values lying between 0 and 1.
* NSP: NSP has several outliers above and below the whiskers, while most observations are concentrated around 1.


```python
#Histogram
df.hist(figsize=(11,5), bins=20)
plt.tight_layout()
plt.show()
```

* DS: DS is extremely right-skewed, with nearly all values equal to 0.
* DP: DP is extremely right-skewed, with almost all observations concentrated at 0.
* ASTV: ASTV is roughly bell-shaped, with most values between 20 and 80.

```python
#Frequency of Categorical Variable (NSP)
plt.figure(figsize=(12,6))
df["NSP"].value_counts().plot(kind="bar")
plt.title("NSP Value Counts")
plt.xlabel("NSP")
plt.ylabel("Count")
plt.show()
#The provided dataset contains many unique floating-point values in the NSP column,so the resulting bar chart shows all distinct 
#values present in the dataset.
```

NSP is concentrated mainly at 1, with some observations 2 adn 3.

```python
#Correlation Heatmap
plt.figure(figsize=(12,6))
sns.heatmap(df.corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

* Most variables have weak corelation (between -0.3:0.3), means they are independent of one another.
* No strong corelations are present among the variables(>0.6)

```python
plt.figure(figsize=(12,6))
plt.scatter(df["LB"], df["ASTV"], alpha=0.7)
plt.xlabel("LB")
plt.ylabel("ASTV")
plt.title("Scatter Plot: LB vs ASTV")
plt.grid(True)
plt.show()
```

* Most data points are concentreted between 120-150 of LB and 20-70 of ASTV.
* It indicates weak positive corelation between ASTV and LB(also beacuse data points are widely scattered)

```python
#Pairplot
sns.pairplot(df[["LB","AC","UC","ASTV","NSP"]],hue="NSP")
plt.show()
```

* The diagonal plots show the distribution of each selected variable, while the off-diagonal plots show pairwise relationships between the variables.
* Most observations are concentrated within a limited range, with a few noticeable outliers across several variable pairs.
* No strong linear relationship is observed between the selected variables, as the data points are widely scattered.
* The NSP classes overlap considerably, meaning that LB, AC, UC, and ASTV alone do not clearly separate the different NSP categories.

### Conclusion

* Most numerical variables showed meaningful distributions while few contained outliers and were positively skewed.
* Corelation analysis showed that few pairs had moderate postive corelation while others had weak corelation.
* With help of scatter and pair plot we noticed few variables had weak corelation and NSP classes overlap considerably making difficult
  to disinguish using only features
