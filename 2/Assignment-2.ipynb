### Given Data


Durability of 15 print-heads(in millions of characters):
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29

Here, 
* Sample size: n = 15
* Sample mean: x-bar = 1.2387
* Sample standard deviation: s = 0.1932

### 99% Confidence Interval Using Sample Standard Deviation

```python
#Since population standard deviation is unknown and sample size is small(n=15<30), we will use t-distribution to calculate the 
#confidence interval for the mean.

import numpy as np
from scipy import stats

data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])

n = len(data)
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1) 
confidence = 0.99
alpha = 1 - confidence

df = n - 1  # degrees of freedom
t_crit = stats.t.ppf(1-alpha/2,df)  #t-critical value
stderr_t = sample_std / np.sqrt(n)   #standard error
margin_t = t_crit * stderr_t     #margin of error

#Confidence Intervals:
lt_t = sample_mean - margin_t #Lower limit
ut_t = sample_mean + margin_t #Upper limit
```

### Confidence Interval using known population Standard Deviation

```python
population_std = 0.2
z_crit = stats.norm.ppf(1-alpha/2)  #z-critical value
stderr_z = population_std / np.sqrt(n)   #standard error
margin_z = z_crit * stderr_z     #margin of error

#Confidence Intervals:
lt_z = sample_mean - margin_z #Lower limit
ut_z = sample_mean + margin_z #Upper limit
```

### Results

```python
print("Part (a):")
print("99% CI (t-distribution): (" + str(round(lt_t, 4)) + ", " + str(round(ut_t, 4)) + ")")

print("\nPart (b):")
print("99% CI (z-distribution): (" + str(round(lt_z, 4)) + ", " + str(round(ut_z, 4)) + ")")
```
