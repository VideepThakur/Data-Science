# Association Rules

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_excel("Online retail.xlsx", header=None)
df.head()
```

```python
#Data Preprocessing
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
transactions = []
for row in df[0]:
    items = row.split(",")
    items = [item.strip() for item in items]
    transactions.append(items)

print("Total Transactions:", len(transactions))
```

```python
# One-Hot Encoding
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)
basket = pd.DataFrame(te_data, columns=te.columns_)
basket.head()
```

```python
#Applying Apriori
frequent_items = apriori(basket,min_support=0.03,use_colnames=True)
frequent_items.sort_values(by="support", ascending=False).head(10)
```

```python
#Generating Association Rules
rules = association_rules(frequent_items,metric="confidence",min_threshold=0.3) #Atleast 30% confidence
rules = rules.sort_values(by="lift",ascending=False)
rules[['antecedents','consequents','support','confidence','lift']].head(20)
```

```python
print("Total Rules:", len(rules))
print("\nTop Frequent Itemsets")
print(frequent_items.sort_values(by="support", ascending=False).head(10))

print("\nTop Association Rules")
print(rules[['antecedents','consequents','support','confidence','lift']].head(10))
```

### Observations

* High support means items appear frequently.
* High confidence means the rule is reliable.
* Lift greater than 1 indicates a positive association.
* These rules help businesses recommend products and improve sales.


### Interview Questions

(Q) What is Lift and why is it important in Association Rules?

* Lift measures how strongly two items are associated compared to if they were independent.
* Lift(A->B) = Confidence(A->B)/Support(B)
* Lift > 1: Positive association between items. Lift = 1: No association. Lift < 1: Negative association.

(Q) What are Support and Confidence? How do you calculate them?
* Support measures how frequently an itemset appears in the dataset.
* Support(A->B) = (Transactions containing both A & B) / Total Transactions
* Confidence measures how often item B appears when item A is present.
* Confidence(A->B) = (Support A union B) / Support(A)

(Q) What are some limitations or challenges of Association Rule Mining?
* It can generate a very large number of rules, making analysis difficult.
* Choosing appropriate minimum support and confidence values can be challenging.
* It requires high computational time for large datasets.
