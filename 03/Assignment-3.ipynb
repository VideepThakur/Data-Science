### Prime Number

```python
import math

def isPrime(n):
    if n<=1: #if it is less than or equal to 1, it is not prime
        return False
    if n==2: #if it is 2, it is prime
        return True
    if n%2==0: #if it is even, it is not prime
        return False
    for i in range(2,int(math.sqrt(n))+1): #if it is odd, checking if it is divisible by any number from 2 to the square root of n
        if n%i==0:
            return False #it is not prime
    return True

n = int(input("Enter a number: "))
if isPrime(n):
    print("It is a prime number.")
else:
    print("It is not a prime number.")
```

### Product of Random Numbers

```python
import random

a = random.randint(1,10) #random number between 1 and 10
b = random.randint(1,10) #random number between 1 and 10
c = a*b

d = int(input("What is " + str(a) + " x " + str(b) + "? "))

if d == c:
    print("Correct!")
else:
    print("Incorrect. The correct answer is", c)
```

### Word Counter

```python
s = str(input("Enter a string: "))

words = s.split()
arr = {}

for i in words:
    if i in arr:
        arr[i]+=1
    else:
        arr[i]=1

for i in arr:
    print(i,":",arr[i])
```

### Palindrome

```python
def is_palindrome(text):
    #convert to lowercase and remove spaces
    text=text.lower().replace(" ","")
    #check if the text is the same forwards and backwards
    if text == text[::-1]: #text[::-1] reverses the string
        return True
    else:
        return False

word=input("Enter a word: ")
if is_palindrome(word):
    print(True)
else:
    print(False)
```
